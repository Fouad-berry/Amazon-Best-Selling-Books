# 01 — Exploratory Data Analysis
Shape, distributions, nulls, correlations — Amazon Best-Selling Books.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', palette='muted')
pd.set_option('display.max_columns', None)

df = pd.read_csv('../data/raw/amazon_bestselling_books.csv')
print(f'Shape: {df.shape}')
df.head()
```

```python
print(df.isnull().sum())
print()
print(df.dtypes)
```

```python
df.describe().T
```

```python
# Category & format distributions
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
df['Category'].value_counts().plot(kind='bar', ax=axes[0], color='steelblue', title='Category')
df['Format'].value_counts().plot(kind='bar', ax=axes[1], color='tomato', title='Format')
df['Sub-Genre'].value_counts().head(12).plot(kind='barh', ax=axes[2], color='seagreen', title='Top 12 Sub-Genres')
for ax in axes: ax.tick_params(axis='x', rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Correlation heatmap
num_cols = ['Rank', 'Price (USD)', 'Rating', 'Reviews', 'Weeks on List', 'Year Published', 'Amazon BSR']
plt.figure(figsize=(10, 7))
sns.heatmap(df[num_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlation Matrix')
plt.tight_layout()
plt.show()
```

```python
# Price & rating distributions
fig, axes = plt.subplots(1, 3, figsize=(16, 4))
df['Price (USD)'].hist(bins=30, ax=axes[0], color='steelblue')
axes[0].set_title('Price Distribution')
df['Rating'].hist(bins=20, ax=axes[1], color='tomato')
axes[1].set_title('Rating Distribution')
df['Reviews'].apply(lambda x: x+1).hist(bins=30, ax=axes[2], color='seagreen', log=True)
axes[2].set_title('Reviews (log scale)')
plt.tight_layout()
plt.show()
```
